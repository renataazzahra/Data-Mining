# Pengecekan Data

```python
x_train = datatrain.drop(columns= ['quality', 'Id'])
y_train = datatrain.quality
```

```python
x_train
```

```python
y_train
```

split datatrain menjadi data training dan data validasi. kemudian kita gunakan test_size 0,1, artinya perbandingan antara data training dan data validasi yaitu 90:10 karena data yang dipunyai relatif kecil, yaitu kurang dari 1000 data. stratify digunakan supaya distribusi label tetap seimbang antara data training dan data validasi

```python
from sklearn.model_selection import train_test_split

x_tr, x_val, y_tr, y_val = train_test_split(
    x_train, y_train, test_size=0.1, random_state=1, stratify=y_train
)
```

Selanjutnya, kita lakukan scaling pada data x_tr dan x_val karena nilai nilai dalam masing masing variabelnya sangat bervariasi.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_tr_scaled = scaler.fit_transform(x_tr)
x_val_scaled = scaler.transform(x_val)
```

  Langkah selanjutnya yaitu saya menguji menggunakan beberapa model klasifikasi untuk menentukan mana model yang terbaik

```python
import statsmodels.api as sm
```

# DECISION TREE

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import preprocessing
```

```python
classifier = DecisionTreeClassifier()
classifier.fit(x_tr_scaled, y_tr)
```

```python
max_depth_tree_list = np.arange(1, 20, 2)
tree_params_list = {
    'max_depth': max_depth_tree_list,
    'criterion': ['log_loss', 'gini', 'entropy'],
}
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold

grid_search = GridSearchCV(
    estimator= DecisionTreeClassifier(),
    param_grid= tree_params_list,
    cv= StratifiedKFold(n_splits= 5),
    scoring= 'f1_macro'
)
grid_search.fit(x_tr_scaled, y_tr)
```
